```python
def extract_embedding(image_path):

    # Ouverture de l'image JPEG
    image = read_image(str(image_path))


    # Conversion du Tensor en image exploitable par torchvision
    # read_image retourne déjà un Tensor
    # On convertit les valeurs en float entre 0 et 1
    image = image.float() / 255.0


    # Application du preprocessing ResNet
    image = transform(image)


    # Ajout d'une dimension batch
    # ResNet attend :
    # (nombre_images, canaux, hauteur, largeur)
    image = image.unsqueeze(0)


    # Désactivation du calcul des gradients
    # Gain de mémoire et de temps
    with torch.no_grad():

        # Extraction des caractéristiques
        embedding = feature_extractor(image)


    # Suppression des dimensions inutiles
    # Passage de (1,2048,1,1) à (2048)
    embedding = embedding.squeeze()


    # Conversion en tableau numpy
    return embedding.numpy()
```

```python
import copy
# ============================================================
# GRID SEARCH MANUEL - MODELE SUPERVISE
# Hyperparamètres : dropout et weight_decay
# ============================================================

# Valeurs testées
dropout_values = [0.0, 0.3, 0.5]

weight_decay_values = [0.0, 1e-4, 1e-3]


# Dictionnaire pour stocker les résultats
grid_results = []

# Meilleur score
best_f1 = 0.0
best_dropout = None
best_weight_decay = None


# ============================================================
# Boucle sur les valeurs de dropout
# ============================================================

for dropout in dropout_values:

    # ========================================================
    # Boucle sur les valeurs de weight_decay
    # ========================================================

    for weight_decay in weight_decay_values:

        print("\n" + "=" * 60)
        print(f"Dropout = {dropout} | Weight decay = {weight_decay}")
        print("=" * 60)


        # ----------------------------------------------------
        # Création d'un nouveau ResNet50
        # ----------------------------------------------------

        model_grid = models.resnet50(
            weights=models.ResNet50_Weights.DEFAULT
        )


        # ----------------------------------------------------
        # Gel de toutes les couches pré-entraînées
        # ----------------------------------------------------

        for param in model_grid.parameters():
            param.requires_grad = False


        # ----------------------------------------------------
        # Remplacement de la couche finale
        # avec Dropout + couche de classification
        # ----------------------------------------------------

        num_features = model_grid.fc.in_features

        model_grid.fc = torch.nn.Sequential(
            torch.nn.Dropout(p=dropout),
            torch.nn.Linear(num_features, 2)
        )


        # ----------------------------------------------------
        # Fonction de perte
        # ----------------------------------------------------

        criterion_grid = torch.nn.CrossEntropyLoss()


        # ----------------------------------------------------
        # Optimiseur
        # weight_decay = régularisation L2
        # ----------------------------------------------------

        optimizer_grid = torch.optim.Adam(
            filter(
                lambda p: p.requires_grad,
                model_grid.parameters()
            ),
            lr=0.001,
            weight_decay=weight_decay
        )


        # ----------------------------------------------------
        # Entraînement
        # ----------------------------------------------------

        num_epochs_grid = 10

        model_grid.train()

        for epoch in range(num_epochs_grid):

            running_loss = 0.0

            for images, labels in train_loader_aug:

                # Remise à zéro des gradients
                optimizer_grid.zero_grad()

                # Prédictions
                outputs = model_grid(images)

                # Calcul de la loss
                loss = criterion_grid(outputs, labels)

                # Rétropropagation
                loss.backward()

                # Mise à jour des poids
                optimizer_grid.step()

                # Accumulation de la loss
                running_loss += loss.item()


            # Affichage de la loss
            print(
                f"Époque {epoch+1}/{num_epochs_grid} - "
                f"Loss : {running_loss / len(train_loader_aug):.4f}"
            )


        # ====================================================
        # EVALUATION SUR LA VALIDATION
        # ====================================================

        model_grid.eval()

        y_true_grid = []
        y_pred_grid = []


        with torch.no_grad():

            for images, labels in validation_loader:

                # Prédictions
                outputs = model_grid(images)

                # Classe prédite
                predictions = torch.argmax(outputs, dim=1)

                # Sauvegarde
                y_true_grid.extend(labels.numpy())
                y_pred_grid.extend(predictions.numpy())


        # ----------------------------------------------------
        # Calcul du F1-score
        # pos_label=0 = Cancer
        # ----------------------------------------------------

        f1_grid = f1_score(
            y_true_grid,
            y_pred_grid,
            pos_label=0
        )


        # Accuracy pour information
        accuracy_grid = accuracy_score(
            y_true_grid,
            y_pred_grid
        )


        # ----------------------------------------------------
        # Sauvegarde des résultats
        # ----------------------------------------------------

        grid_results.append({
            "dropout": dropout,
            "weight_decay": weight_decay,
            "f1": f1_grid,
            "accuracy": accuracy_grid
        })


        # ----------------------------------------------------
        # Affichage
        # ----------------------------------------------------

        print(
            f"Validation - "
            f"Accuracy : {accuracy_grid:.3f} | "
            f"F1-score : {f1_grid:.3f}"
        )


        # ----------------------------------------------------
        # Recherche du meilleur modèle
        # ----------------------------------------------------

        if f1_grid > best_f1:

            best_f1 = f1_grid
            best_dropout = dropout
            best_weight_decay = weight_decay

            # Sauvegarde des poids du meilleur modèle
            best_model_state = copy.deepcopy(model_grid.state_dict())


# ============================================================
# RESULTAT FINAL DU GRID SEARCH
# ============================================================

print("\n" + "=" * 60)
print("MEILLEURE COMBINAISON")
print("=" * 60)

print(f"Dropout       : {best_dropout}")
print(f"Weight decay  : {best_weight_decay}")
print(f"F1-score      : {best_f1:.3f}")

### Affichage de tous les résultats

# Conversion des résultats en DataFrame
grid_results_df = pd.DataFrame(grid_results)

# Tri du meilleur au moins bon
grid_results_df = grid_results_df.sort_values(
    by="f1",
    ascending=False
)

print(grid_results_df)
```
